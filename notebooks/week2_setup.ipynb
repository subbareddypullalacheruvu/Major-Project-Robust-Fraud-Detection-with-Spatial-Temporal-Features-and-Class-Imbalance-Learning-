```python
import pandas as pd

ulb = pd.read_csv('data/ulb/creditcard.csv')
print("ULB shape:", ulb.shape)
print(ulb['Class'].value_counts())

counts = ulb['Class'].value_counts()
ulb_fraud_rate = counts[1] / counts.sum()
ulb_ratio = counts[0] / counts[1]
print(f"\nFraud rate: {ulb_fraud_rate:.4%}")
print(f"Imbalance ratio (legit:fraud): {ulb_ratio:.1f} : 1")
```

```python
sparkov_train = pd.read_csv('data/sparkov/fraudTrain.csv', index_col=0)
sparkov_test = pd.read_csv('data/sparkov/fraudTest.csv', index_col=0)
sparkov = pd.concat([sparkov_train, sparkov_test], ignore_index=True)

print("Sparkov shape:", sparkov.shape)
print(sparkov['is_fraud'].value_counts())

counts = sparkov['is_fraud'].value_counts()
sparkov_fraud_rate = counts[1] / counts.sum()
sparkov_ratio = counts[0] / counts[1]
print(f"\nFraud rate: {sparkov_fraud_rate:.4%}")
print(f"Imbalance ratio (legit:fraud): {sparkov_ratio:.1f} : 1")
```
